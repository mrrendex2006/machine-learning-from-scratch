```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
```

```python
df=pd.read_csv('Phishing_URL_Dataset.csv')
df.head()
```

** Only selecting the important feature :**

```python
print("Total row:",len(df))
print("number of repeatation of url in dataset:",df['URL'].duplicated().sum())
print("unique URL:",df['URL'].nunique())
```

```python
# checking if duplicated url have the same label.
duplicate_url=df[df['URL'].duplicated(keep=False)].sort_values('URL')
print(
    duplicate_url.groupby('URL')['label'].nunique().value_counts()
)
```

```python
corr_matrix = df.corr(numeric_only=True)[['label']]
print(type(corr_matrix))
print(corr_matrix.shape)

#ploting heatmap 

plt.figure(figsize=(20,18))
sns.heatmap(corr_matrix,annot=True, cmap='coolwarm',fmt='.2f',linewidths=0.5)
plt.title('correlation matrix of features')
plt.show()
```

```python
x_url=[
    "URLLength",
    "DomainLength",
    "IsDomainIP",
    "TLDLength",
    "NoOfSubDomain",
    "HasObfuscation",
    "NoOfObfuscatedChar",
    "ObfuscationRatio",
    "NoOfLettersInURL",
    "LetterRatioInURL",
    "NoOfDegitsInURL",
    "NoOfQMarkInURL",
    "NoOfAmpersandInURL",
    "NoOfOtherSpecialCharsInURL",
    "SpacialCharRatioInURL",
    "IsHTTPS"
] # list of features we will use to train our model 
target=df["label"]
```

```python
X=df[x_url]
y=target
print("shape of the feature variable X:",X.shape)
print("shape of the feature variable y:",y.shape)
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,y, test_size=0.30,random_state=42,stratify=y)
print("shape of traing data :",x_train.shape)
print("shape of test data :",x_test.shape)
```

```python
x_train.columns
```

```python
scaler=StandardScaler()
x_train_sc=scaler.fit_transform(x_train)
x_test_sc=scaler.transform(x_test)
```

```python
lr=LogisticRegression()
lr.fit(x_train_sc,y_train)
```

```python
y_pred=lr.predict(x_test_sc)
```

```python
cm=confusion_matrix(y_test,y_pred)
print("confusion_matrix:\n",cm)

print("Classification report :\n",classification_report(y_test,y_pred))
```

```python
score=accuracy_score(y_test,y_pred)
print("score of logistic regression:",score)
```

```python
knn=KNeighborsClassifier()
knn.fit(x_train_sc,y_train)
```

```python
y_pred=knn.predict(x_test_sc)
```

```python
print("confusion_matrix:\n",cm)

print("Classification report :\n",classification_report(y_test,y_pred))
```

```python
score=accuracy_score(y_test,y_pred)
print("score",score)
```
