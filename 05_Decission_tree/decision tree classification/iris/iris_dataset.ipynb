```python
import pandas as pd 
import matplotlib.pyplot as plt
%matplotlib inline 
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
```

```python
df.head()
```

```python
#dependent and indipendent features 
X=df.iloc[:,:-1]
y=df['species']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.33,random_state=42)

```

```python
from  sklearn.tree import DecisionTreeClassifier
treemodel=DecisionTreeClassifier(max_depth=2)#we do not need to split to the last leaf node 
                          # already seen without hyperparamenter max_depth 
treemodel.fit(X_train,y_train)
```

here providing hyperparameter is known as post prunning 

```python
tree_predict=treemodel.predict(X_test)
```

```python
from sklearn import tree
plt.figure(figsize=(10,10))
tree.plot_tree(treemodel,filled=True)
```

```python
from sklearn.metrics import classification_report,accuracy_score
print("---------Accuracy :",accuracy_score(y_test,tree_predict))
print("---------classification report :-\n",classification_report(y_test,tree_predict))
```

now lets check for best parameters best fit for our iris dataset for decision tree . we will apply gridsearch cv and try pre prunning by providing parameter before traing our model

```python
parameter={
    'criterion':['gini','entropy','log_loss'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4],
    'max_features':['sqrt','log2'],
    
}
```

```python
from sklearn.model_selection import GridSearchCV
cv=GridSearchCV(treemodel,param_grid=parameter,cv=5,scoring='accuracy')
```

```python
cv.fit(X_train,y_train)
```

```python
cv.best_params_ # it gives the best combination of paramenter that best fit the data 
```

```python
y_pred=cv.predict(X_test)
```

```python

print("---------Accuracy :",accuracy_score(y_test,y_pred))
print("---------classification report :-\n",classification_report(y_test,y_pred))
```
