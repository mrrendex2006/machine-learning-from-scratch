```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
)
```

```python
df = pd.read_csv("Summarised_Recalculated_Coffee_Waste_Physiology_Dataset.csv")
df.head()
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
features=["SM","ST","Tair",
          "pH","SBD","SOM",
          "SAK","SOP","SAN"]

Target="Treatment"

X=df[features].values
y=df[Target].values
```

logistics regression needs integer class lebels . lablEncoders assign C=0 ,L=1 ,LP=2 ,P=3

```python

le=LabelEncoder()
y_enc=le.fit_transform(y)
class_name=le.classes_
```

```python

X_train, X_test, y_train, y_test = train_test_split(
    X, y_enc, test_size=0.25, random_state=42, stratify=y_enc
)
print(f"Train: {len(X_train)} rows | Test: {len(X_test)} rows")
print(f"Test class distribution: { {cls: int((y_test==i).sum()) for i,cls in enumerate(class_name)} }")
print()
```

```python

scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s  = scaler.transform(X_test)
 
print("Feature means after scaling (should be ~0):")
print(" ", np.round(X_train_s.mean(axis=0), 3))
print("Feature stds after scaling (should be ~1):")
print(" ", np.round(X_train_s.std(axis=0), 3))
print()
```

```python

model = LogisticRegression(
    C=1.0,          # L2 regularisation (default when penalty not set)
    solver="lbfgs",
    max_iter=1000,
    random_state=42,
)
 
model.fit(X_train_s, y_train)
y_pred = model.predict(X_test_s)
y_prob = model.predict_proba(X_test_s)
```

```python
print("=" * 58)
print("CLASSIFICATION REPORT")
print("=" * 58)
print(classification_report(y_test, y_pred, target_names=class_name, zero_division=0))
 
print("What each column means:")
print("  Precision: of all samples predicted as class X,")
print("             how many actually were X?")
print("  Recall   : of all true class X samples,")
print("             how many did the model catch?")
print("  F1-score : harmonic mean of precision and recall.")
print("             Use this when classes are imbalanced.")
print()
```
