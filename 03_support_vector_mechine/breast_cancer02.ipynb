solving classification problem on dataset brest cancer using support vector 


```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt  
import seaborn as sns 
from sklearn.datasets import load_breast_cancer
```

```python
cancer=load_breast_cancer()

df=pd.DataFrame(cancer.data,columns=cancer.feature_names)
df['target']=cancer.target
x=cancer.data
y=cancer.target
```

```python
print("feature are :",cancer.feature_names)
print("targest are :",cancer.target_names)
```

```python
df.shape
```

```python
corr_matrix=df.corr(numeric_only=True) # this will work as it drops columns have non numeric data
print(corr_matrix)
plt.figure(figsize=(20,20))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('correlation matrix of features')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
x_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=40)
```

```python
from sklearn.svm import SVC
svm=SVC()
svm.fit(x_train,y_train)
print("model trained succesfully ")
```

```python
y_pred=svm.predict(X_test)
print(classification_report(y_test,y_pred))
```
